#### 1. NumPy’s core object is the ndarray (N-dimensional array). Unlike standard Python lists, NumPy arrays allow you to do math on entire blocks of numbers instantly.


NumPy arrays use vectorization to perform mathematical operations on entire blocks of data instantly without writing a single for loop.

example: numpy_array = np.array([1, 2, 3, 4, 5])


### Why NumPy is Superior for Math

- No Explicit Loops: You write direct mathematical expressions like array * 2 or array1 + array2.

- C-Engine Speed: The iteration happens in optimized, pre-compiled C code under the hood, making it up to 100 times faster on massive datasets.

- Contiguous Memory: NumPy stores raw numbers right next to each other in memory. Standard Python lists store scattered pointers to objects, which slows down your CPU.

- Element-Wise Operations: Operations are automatically applied to every individual element in the block.



-- Vectorization means performing a mathematical operation on an entire array of numbers all at once, rather than looping through each item one by one.



How Vectorization Works Under the Hood
Vectorization skips Python's slow internal machinery using two main core mechanisms:

1. SIMD Hardware Instructions: Modern computer processors have a feature called SIMD (Single Instruction, Multiple Data). NumPy utilizes highly optimized compiled libraries (like BLAS or LAPACK) that talk directly to this hardware. Instead of telling your CPU "multiply this number," it tells the CPU "multiply this entire block of 4, 8, or 16 numbers in a single clock cycle."

2. Eliminating Python Type-Checking Overhead: Because standard Python lists can hold a mix of anything—integers, strings, or even other lists—Python must pause at every single element during a loop to check its data type and look up the correct math operation.NumPy arrays are homogeneous (every single element must be the exact same data type, like 64-bit float). Because the data type is completely predictable, NumPy bypasses all type-checking and safety checks, executing the math at raw hardware speeds


```Here is what happens behind the scenes for both approaches:python# --- WHAT PYTHON DOES (Scalar Loop) ---

# For item 2: Check type -> It's an int -> Multiply -> Store pointer
for x in python_list:
    result.append(x * 2)

# --- WHAT NUMPY DOES (Vectorized) ---
# All ints? Yes. CPU, execute SIMD multiplication on this whole block now.
numpy_array * 2```

# Handling Missing Data

Real world datasets are messy, Numpy uses a special floating point value called nan ( Not a number) to represent missing information. Because standard math functions fail if a nan is present. we must use specific 'nan safe' operations

````python
sensor_data = np.array([10.5 , 12.1 , np.nan , 12.3])

# mean
np.mean(sensor_data) # Fails
````

````python
np.nanmean(sensor_data)
````

````python
# Data Transformation ( Log , Exponent)

np.exp(sensor_data)
````

````python
# 1. Define the complete dataset
data = {
    'Store': ['Downtown', 'Westside', 'Downtown', 'Westside', 'Downtown'],
    'Sales': [100, 150, 200, 120, 180],  # Added mock sales numbers
    'Customers': [50, 40, 65, 42, 48]
}
df = pd.DataFrame(data)

# 2. Apply different mathematical functions to different columns
store_summary = df.groupby('Store').agg({
    'Sales': ['sum', 'mean'], # Total sales and average sale per store
    'Customers': 'max'        # The highest single-day customer count
})

# 3. View the final aggregated table
print(store_summary)
````

````python
df = pd.DataFrame({
    'Employee': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['IT', 'IT', 'Sales', 'Sales'],
    'Salary': [80000, 50000, 70000, 40000]
})

# Calculate the department average, but broadcast it back to match original rows
df['Dept_Avg_Salary'] = df.groupby('Department')['Salary'].transform('mean')

# Calculate the deviation feature for a machine learning model
df['Salary_Deviation'] = df['Salary'] - df['Dept_Avg_Salary']
print(df)
````
